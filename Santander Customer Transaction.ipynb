###Split

```python
X = train.drop(['ID_code', 'target'], axis=1)
y = train['target']
```

### Baseline Models with Outliers

```python
# CV Setup

from sklearn.model_selection import cross_val_score, StratifiedKFold

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

scores = cross_val_score(rf, X, y, cv=cv, scoring='roc_auc')

print("RandomForest — ROC-AUC per fold:", scores)
print("Mean ROC-AUC:", scores.mean())

# RandomForest — ROC-AUC per fold: [0.8350142  0.84045388 0.8291124  0.83279035 0.83779867]
# Mean ROC-AUC: 0.835033899616889
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_val_score

nb = GaussianNB()

scores = cross_val_score(nb, X, y, cv=cv, scoring='roc_auc')

print("GaussianNB — ROC-AUC per fold:", scores)
print("Mean ROC-AUC:", scores.mean())

# GaussianNB — ROC-AUC per fold: [0.89069373 0.88900929 0.88209134 0.88896748 0.89109282]
# Mean ROC-AUC: 0.8883709344999264
```

```python
from xgboost import XGBClassifier
from sklearn.model_selection import cross_val_score

xgb = XGBClassifier(
    tree_method="hist",
    eval_metric="logloss",
    random_state=42
)

scores = cross_val_score(xgb, X, y, cv=cv, scoring='roc_auc')

print("XGBoost — ROC-AUC per fold:", scores)
print("Mean ROC-AUC:", scores.mean())

# XGBoost — ROC-AUC per fold: [0.858824   0.86061547 0.85170603 0.85824797 0.8609689 ]
# Mean ROC-AUC: 0.8580724760093028
```

```python
from lightgbm import LGBMClassifier
from sklearn.model_selection import cross_val_score

lgb = LGBMClassifier(random_state=42)

scores = cross_val_score(lgb, X, y, cv=cv, scoring='roc_auc')

print("LightGBM — ROC-AUC per fold:", scores)
print("Mean ROC-AUC:", scores.mean())

# LightGBM — ROC-AUC per fold: [0.86584068 0.8694356  0.85916321 0.86601633 0.86558445]
# Mean ROC-AUC: 0.8652080552839129
```

```python
# CatBoost ROC-AUC with fold-by-fold scores

from catboost import CatBoostClassifier
from sklearn.model_selection import cross_val_score

cat = CatBoostClassifier(
    iterations=300,
    depth=6,
    learning_rate=0.1,
    eval_metric='AUC',
    verbose=False,
    random_state=42
)

scores = cross_val_score(cat, X, y, cv=cv, scoring='roc_auc')

print("CatBoost — ROC-AUC per fold:", scores)
print("Mean ROC-AUC:", scores.mean())

# CatBoost — ROC-AUC per fold: [0.88660179 0.88834447 0.88000804 0.88597168 0.88824158]
# Mean ROC-AUC: 0.8858335117868033
```

###Hyperparameter tuning

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# GaussianNB pipeline with scaling (very important!)
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("nb", GaussianNB())
])

# var_smoothing influences numerical stability
param_grid = {
    "nb__var_smoothing": [1e-09, 1e-08, 1e-07, 1e-06, 1e-05, 1e-04, 1e-03]
}

grid = GridSearchCV(
    estimator=pipe,
    param_grid=param_grid,
    scoring="roc_auc",
    cv=cv,
    n_jobs=-1
)

grid.fit(X, y)

print("Best ROC-AUC:", grid.best_score_)
print("Best params:", grid.best_params_)
print("\nFold-by-fold ROC-AUC:", grid.cv_results_['mean_test_score'])
```

```python
import optuna
from xgboost import XGBClassifier
from sklearn.model_selection import cross_val_score

def objective(trial):
    # Hyperparameters to tune
    params = {
        "tree_method": "hist",
        "eval_metric": "logloss",
        "random_state": 42,
        "n_estimators": trial.suggest_int("n_estimators", 200, 800),
        "max_depth": trial.suggest_int("max_depth", 3, 10),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
        "subsample": trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.5, 1.0),
        "gamma": trial.suggest_float("gamma", 0.0, 5.0),
        "min_child_weight": trial.suggest_int("min_child_weight", 1, 10)
    }

    model = XGBClassifier(**params)

    # Cross-validation ROC-AUC
    score = cross_val_score(model, X, y, cv=cv, scoring="roc_auc").mean()
    return score


# Create study
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=40, show_progress_bar=True)


#  Trial 13 finished with value: 0.8924144955542964 and parameters:
# {'n_estimators': 769, 'max_depth': 3, 'learning_rate': 0.10201756699194674, 'subsample': 0.6842310031324624, 'colsample_bytree': 0.6652004724194502, 'gamma': 0.10262057525440017, 'min_child_weight': 7}. Best is trial 13 with value: 0.8924144955542964.
```

```python
import optuna
from catboost import CatBoostClassifier
from sklearn.model_selection import cross_val_score

def objective(trial):
    params = {
        "iterations": trial.suggest_int("iterations", 200, 800),
        "depth": trial.suggest_int("depth", 4, 10),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
        "l2_leaf_reg": trial.suggest_float("l2_leaf_reg", 1, 10),
        "border_count": trial.suggest_int("border_count", 32, 255),
        "random_strength": trial.suggest_float("random_strength", 0.5, 5.0),
        "bagging_temperature": trial.suggest_float("bagging_temperature", 0.0, 5.0),
        "eval_metric": "AUC",
        "verbose": False,
        "random_state": 42,
        # Important to avoid GPU errors if no GPU
        "task_type": "CPU"
    }

    model = CatBoostClassifier(**params)

    score = cross_val_score(model, X, y, cv=cv, scoring="roc_auc").mean()
    return score


study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=40, show_progress_bar=True)

# Trial 34 finished with value: 0.8944462076633435 and parameters:
# {'iterations': 572, 'depth': 6, 'learning_rate': 0.12128509650816555, 'l2_leaf_reg': 6.582277969773736, 'border_count': 141, 'random_strength': 3.4620331101513475, 'bagging_temperature': 3.3454297284366383}. Best is trial 34 with value: 0.8944462076633435.
```

###Ensemble Stacking

```python
from catboost import CatBoostClassifier

catboost_model = CatBoostClassifier(
    iterations=572,
    depth=6,
    learning_rate=0.12128509650816555,
    l2_leaf_reg=6.582277969773736,
    border_count=141,
    random_strength=3.4620331101513475,
    bagging_temperature=3.3454297284366383,
    eval_metric='AUC',
    random_seed=42,
    verbose=False
)

catboost_model.fit(X, y)
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=769,
    max_depth=3,
    learning_rate=0.10201756699194674,
    subsample=0.6842310031324624,
    colsample_bytree=0.6652004724194502,
    gamma=0.10262057525440017,
    min_child_weight=7,
    objective='binary:logistic',
    eval_metric='auc',
    random_state=42
)

xgb_model.fit(X, y)
```

```python
from sklearn.naive_bayes import GaussianNB

gnb_model = GaussianNB(var_smoothing=1e-9)
gnb_model.fit(X, y)
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression

estimators = [
    ('cat', catboost_model),
    ('xgb', xgb_model),
    ('gnb', gnb_model)
]

stack = StackingClassifier(
    estimators=estimators,
    final_estimator=LogisticRegression(max_iter=500),
    cv=5,
    stack_method='predict_proba'
)

stack.fit(X, y)
```

```python
from sklearn.metrics import roc_auc_score

y_pred_stack = stack.predict_proba(X)[:, 1]
auc_stack = roc_auc_score(y, y_pred_stack)

print("Stacking ROC-AUC:", auc_stack)
```

```python
y_pred_weighted = (
    0.6 * catboost_model.predict_proba(X)[:, 1] +
    0.3 * xgb_model.predict_proba(X)[:, 1] +
    0.1 * gnb_model.predict_proba(X)[:, 1]
)

auc_weighted = roc_auc_score(y, y_pred_weighted)
print("Weighted Ensemble ROC-AUC:", auc_weighted)
```
